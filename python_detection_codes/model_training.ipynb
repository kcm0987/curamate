```python
import pandas as pd
import numpy as np
import re
import torch
import matplotlib.pyplot as plt
import seaborn as sns
import pickle

# Install required packages
print("Installing required packages...")
import subprocess
import sys

def install_packages():
    packages = [
        "nltk",
        "transformers",
        "datasets",
        "torch",
        "scikit-learn"
    ]
    for package in packages:
        subprocess.check_call([sys.executable, "-m", "pip", "install", package])

install_packages()

# Now import the packages that required installation
import nltk
from nltk.corpus import stopwords
from transformers import BertTokenizer, BertForSequenceClassification, Trainer, TrainingArguments
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report, confusion_matrix
from datasets import Dataset

# Download required NLTK resources
nltk.download('stopwords')

# Load the CSV data
data = pd.read_csv("thyroid_data.csv")
print("Original data:")
print(data.head())

# Load stopwords
stop_words = set(stopwords.words('english'))

# Text cleaning function
def clean_text(text):
    # Convert to string and lowercase the text
    text = str(text).lower()

    # Remove any numbers (modify if numbers are important)
    text = re.sub(r'\d+', '', text)

    # Remove special characters, punctuation, and non-alphabetical characters
    text = re.sub(r'[^a-z\s]', '', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # Remove stopwords
    text = ' '.join([word for word in text.split() if word not in stop_words])

    return text

# Apply the cleaning function to the Clinical Notes column
data['Clinic Notes'] = data['Clinic Notes'].apply(clean_text)
print("\nCleaned data:")
print(data.head())

# Check distribution of thyroid diseases
print("\nThyroid Disease distribution:")
print(data['Thyroid Disease'].value_counts())

# Label encoding for 'Thyroid Disease' column
label_encoder = LabelEncoder()
data['Thyroid Disease Encoded'] = label_encoder.fit_transform(data['Thyroid Disease'])

print("\nEncoded data:")
print(data[['Thyroid Disease', 'Thyroid Disease Encoded']].head())

# Split data into train and test
train_texts, test_texts, train_labels, test_labels = train_test_split(
    data['Clinic Notes'],
    data['Thyroid Disease Encoded'],
    test_size=0.2,
    random_state=42
)

# Tokenization using BERT tokenizer
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# Prepare the data for BERT
train_encodings = tokenizer(list(train_texts), padding=True, truncation=True, max_length=512)
test_encodings = tokenizer(list(test_texts), padding=True, truncation=True, max_length=512)

# Convert to Dataset format
train_dataset = Dataset.from_dict({
    'input_ids': train_encodings['input_ids'],
    'attention_mask': train_encodings['attention_mask'],
    'labels': train_labels.tolist()
})

test_dataset = Dataset.from_dict({
    'input_ids': test_encodings['input_ids'],
    'attention_mask': test_encodings['attention_mask'],
    'labels': test_labels.tolist()
})

print("\nSample dataset entry:")
print(train_dataset[0])

# Load model
model = BertForSequenceClassification.from_pretrained(
    'bert-base-uncased',
    num_labels=len(label_encoder.classes_)
)

# Set training arguments
training_args = TrainingArguments(
    output_dir="./thyroid_results",
    evaluation_strategy="epoch",
    save_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=8,  # Reduced batch size for smaller dataset
    per_device_eval_batch_size=8,
    num_train_epochs=10,            # Increased epochs for smaller dataset
    weight_decay=0.01,
    logging_dir="./thyroid_logs",
    logging_steps=5,                # Log more frequently for smaller dataset
    lr_scheduler_type="linear",
    warmup_steps=100,               # Reduced warmup steps for smaller dataset
    load_best_model_at_end=True,
    metric_for_best_model="eval_loss",
    save_total_limit=3,
    gradient_accumulation_steps=2,
    # Added to prevent overfitting on small dataset
    fp16=False,                     # Use if GPU supports it
    seed=42
)

# Trainer
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=test_dataset,
)

# Fine-tuning the model
print("\nTraining the model...")
trainer.train()

# Evaluate the model
print("\nEvaluating the model...")
predictions, labels, _ = trainer.predict(test_dataset)

# Convert predictions to class labels
predicted_labels = np.argmax(predictions, axis=1)

# Make sure the 'label_encoder.classes_' are passed as string labels
target_names = [str(label) for label in label_encoder.classes_]

# Print Classification Report
print("\nClassification Report:")
print(classification_report(test_labels, predicted_labels, target_names=target_names))

# Confusion Matrix
cm = confusion_matrix(test_labels, predicted_labels)

# Plotting the confusion matrix
plt.figure(figsize=(10, 7))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=target_names, yticklabels=target_names)
plt.title('Confusion Matrix')
plt.xlabel('Predicted Labels')
plt.ylabel('True Labels')
plt.show()

# Save the model and tokenizer
print("\nSaving the model...")
trainer.save_model('./thyroid_model')
tokenizer.save_pretrained('./thyroid_model')

# Save the label encoder
with open("thyroid_label_encoder.pkl", 'wb') as f:
    pickle.dump(label_encoder, f)

# Function to make prediction
def predict_thyroid_disease(clinic_note, model, tokenizer, label_encoder):
    # Clean the text
    cleaned_note = clean_text(clinic_note)

    # Tokenize the input note
    inputs = tokenizer(cleaned_note, return_tensors="pt", padding=True, truncation=True, max_length=512)

    # Make prediction
    with torch.no_grad():
        outputs = model(**inputs)
        logits = outputs.logits

    # Get the predicted label
    predicted_label = torch.argmax(logits, dim=1).item()

    # Convert the predicted label to the corresponding disease name
    predicted_disease = label_encoder.inverse_transform([predicted_label])[0]

    return predicted_disease

# Example prediction function
def run_example_predictions():
    # Load the saved model
    model = BertForSequenceClassification.from_pretrained('./thyroid_model')

    # Load the saved tokenizer
    tokenizer = BertTokenizer.from_pretrained('./thyroid_model')

    # Load the label encoder
    with open("thyroid_label_encoder.pkl", 'rb') as f:
        label_encoder = pickle.load(f)

    # Example notes to test
    test_notes = [
        "Patient reports constant fatigue, feeling cold, and constipation with weight gain.",
        "Patient complains of weight loss, irritability, and increased sweating with elevated heart rate.",
        "Patient presents with visible neck swelling and difficulty swallowing."
    ]

    # Make predictions
    for note in test_notes:
        predicted_disease = predict_thyroid_disease(note, model, tokenizer, label_encoder)
        print(f"Note: {note}")
        print(f"Predicted Disease: {predicted_disease}\n")

# Run example predictions
print("\nRunning example predictions...")
run_example_predictions()

# Optional: Create a simple function to predict on new data
def predict_on_new_data(input_note):
    # Load the saved model
    model = BertForSequenceClassification.from_pretrained('./thyroid_model')

    # Load the saved tokenizer
    tokenizer = BertTokenizer.from_pretrained('./thyroid_model')

    # Load the label encoder
    with open("thyroid_label_encoder.pkl", 'rb') as f:
        label_encoder = pickle.load(f)

    # Predict
    predicted_disease = predict_thyroid_disease(input_note, model, tokenizer, label_encoder)

    return predicted_disease

# Example usage of the prediction function
new_note = "Patient describes fatigue, sensitivity to cold, and unexplained weight gain over past few months"
print(f"\nNew patient note: {new_note}")
print(f"Predicted thyroid condition: {predict_on_new_data(new_note)}")
```
